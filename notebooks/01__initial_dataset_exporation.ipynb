# 1. Initial dataset exploration

```python
import pandas as pd
```

```python
# Reading the data from the file
df = pd.read_csv("../data/raw/paranmt_for_detox_500k.tsv", sep="\t", index_col=0)
df.head()
```

```python
# Plot the distribution of text lengths
import matplotlib.pyplot as plt

plt.hist(df["reference"].str.len(), bins=100, label="reference")
plt.hist(df["translation"].str.len(), bins=100, label="translation")
plt.title("Distribution of text lengths")
plt.xlabel("Text length")
plt.ylabel("Count")
plt.yscale("log")
plt.legend()
plt.show()
```
